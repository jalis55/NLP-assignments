```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import re
import math
from tqdm import tqdm
```

```python
driver = webdriver.Chrome()
driver.maximize_window()
```

```python
# base url
driver.get('https://www.daraz.com.bd/routers/')
```

```python
total_item_div = driver.find_element(By.CSS_SELECTOR, "div.M4pDu")
```

```python
total_items=int(re.search(r'\d+', total_item_div.text).group())
total_items
```

```python
items_in_page=driver.find_elements(By.CSS_SELECTOR,'div.qmXQo')
```

```python
item_per_page=len(items_in_page)
item_per_page
```

```python
# total pages
total_pages=math.ceil(total_items/item_per_page)
total_pages
```

```python
def scrape_data(items_in_page):
    item_data = []
    for item_no in range(len(items_in_page)):
        item_key = f'item{item_no+1}'
        item_details = {}
        item_content = items_in_page[item_no].text.split('\n')
        item_name = item_content[0]
        item_price = item_content[1]
        # Check if 'item_content' has at least 3 elements
        if len(item_content) >= 3:
            if re.search(r'sold', item_content[2]):
                total_sold = item_content[2].split(" ")[0]
            else:
                total_sold = None
        else:
            total_sold = None
        
        item_details['item_name'] = item_name
        item_details['item_price'] = item_price
        item_details['total_sold'] = total_sold
        item_data.append({item_key: item_details})
    return item_data

    
```

```python
data=[]
data.append({'page1':scrape_data(items_in_page)})
```

```python

for page in tqdm(range(1,total_pages), desc=f"Scraping data", position=0):
    page_no = f'page{page+1}'
    driver.get(f'https://www.daraz.com.bd/routers/?page={page+1}')
    items_in_page = driver.find_elements(By.CSS_SELECTOR, 'div.qmXQo')
    item_data = scrape_data(items_in_page)
    data.append({page_no: item_data})
```
